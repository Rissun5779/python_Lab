### (b) 利用seaborn 載入 iris 資料，使用 Seaborn 繪製盒狀圖，繪製sepal_length，sepal_width，petal_length, petal_width 其盒狀圖請清楚標註 title 及 x_lable y_label 及單位，各圖請清楚說明你的觀察(寫出觀察心得)

```python
sns.boxplot(x="species", y="sepal_length", data=iris)
plt.title("Iris Sepal Length Boxplot")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
plt.show()
```

```python
sns.boxplot(x="species", y="sepal_width", data=iris)
plt.title("Iris Sepal Width Boxplot")
plt.xlabel("Species")
plt.ylabel("Sepal Width (cm)")
plt.show()
```

```python
sns.boxplot(x="species", y="petal_length", data=iris)
plt.title("Iris Petal Length Boxplot")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.show()
```

```python
sns.boxplot(x="species", y="petal_width", data=iris)
plt.title("Iris Petal Width Boxplot")
plt.xlabel("Species")
plt.ylabel("Petal Width (cm)")
plt.show()
```
