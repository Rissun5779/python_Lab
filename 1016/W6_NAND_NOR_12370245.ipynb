```python
import numpy as np

class Perceptron:
    def __init__(self, input_size, weights, bias):
        if weights is None:
            self.weights = np.ones(input_size)
        else:
            self.weights = weights

        if bias is None:
            self.bias = 0
        else:
            self.bias = bias

    def activation_function(self, x):
        return 1 if x > 0 else 0
    
    def __call__(self, input_data):
        weighted_input = np.dot(input_data, self.weights) + self.bias
        return self.activation_function(weighted_input)    
```

## NAND Gate
### $W_1 = -1,\ W_2 = -1,\ Bias=1.5$ 

```python
weights = np.array([-1, -1])
bias = 1.5
AND_Gate = Perceptron(input_size=2, weights=weights, bias=bias)

input_data = np.array([
    [0, 0], 
    [0, 1], 
    [1, 0], 
    [1, 1]])

for x in input_data:
    output = AND_Gate(x)
    print(f"Input: {x}, Output: {output}")
```

### $W_1 = -0.5,\ W_2 = -0.5,\ Bias=1$ 

```python
weights = np.array([-0.5, -0.5])
bias = 1
AND_Gate = Perceptron(input_size=2, weights=weights, bias=bias)

input_data = np.array([
    [0, 0], 
    [0, 1], 
    [1, 0], 
    [1, 1]])

for x in input_data:
    output = AND_Gate(x)
    print(f"Input: {x}, Output: {output}")
```

## NOR Gate
### $W_1 = -1,\ W_2 = -1,\ Bias=1$ 

```python
weights = np.array([-1, -1])
bias = 1
NOR_Gate = Perceptron(input_size=2, weights=weights, bias=bias)

input_data = np.array([
    [0, 0], 
    [0, 1], 
    [1, 0], 
    [1, 1]])

for x in input_data:
    output = NOR_Gate(x)
    print(f"Input: {x}, Output: {output}")
```

### $W_1 = -0.5,\ W_2 = -0.5,\ Bias=0.5$ 

```python
weights = np.array([-0.5, -0.5])
bias = 0.5
NOR_Gate = Perceptron(input_size=2, weights=weights, bias=bias)

input_data = np.array([
    [0, 0], 
    [0, 1], 
    [1, 0], 
    [1, 1]])

for x in input_data:
    output = NOR_Gate(x)
    print(f"Input: {x}, Output: {output}")
```

## 結論與心得

本次作業透過 Python 物件定義 `Perceptron` 類別，結合 `np.dot` 向量內積與步階激勵函數，成功實現數位邏輯中的兩個基本邏輯閘：NAND 與 NOR。

從參數設計的意義來看，單層感知機的本質是在二維平面中建立一條直線 $W \cdot X + b = 0$ 作為決策邊界。由於 NAND 與 NOR 皆具備一個共同特性：當輸入為 $[0, 0]$ 時輸出為 $1$，而當輸入端出現 $[1, 1]$ 時則輸出為 $0$。因此，兩者的權重 $W$ 皆須設為負值（如 $-1$ 與 $-0.5$）來減少加權總和，並搭配正值的 Bias 來維持正確的激勵輸出。

此外，實驗結果也展示了邊界條件與線性縮放的特性。由於程式中的激勵函數定義為 `x > 0` 才輸出 $1$，當權重固定為 $[-1, -1]$ 時，僅需將 Bias 從 $1.5$（NAND）調至 $1$（NOR），決策邊界便會往原點平移，使 $[0, 1]$ 與 $[1, 0]$ 的加權總和恰好落於 $0$ 而轉為輸出 $0$；同時將權重與偏差值等比例減半，也能得到完全相同的真值表輸出。最後，透過這次作業我得知了既然單層感知機能實現 NAND 與 NOR 這兩個萬用閘，那類似的只要將多個感知機串並聯成多層感知機（MLP），即可組合出如 XOR 等任意複雜的非線性邏輯。
