```python
import os
os.environ["KERAS_BACKEND"] = "torch"

import numpy as np
import pandas as pd
from keras import Sequential
from keras.layers import Dense, Input

np.random.seed(50)
```

```python
!wget -q https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv -O diabetes.csv

data = pd.read_csv("./diabetes.csv")
dataset = data.values
np.random.shuffle(dataset)

X = dataset[:, 0:8]
Y = dataset[:, 8]
```

## 基礎架構
(1) 以 ch5-2-2.py 為基礎，執行訓練測試，結果正確率為何

### 模型
Input (8)-> Dense (10) -> Dense(8) -> Output(1) 

```python
model = Sequential()

# model.add(Input(shape=(8,)))
# model.add(Dense(10, activation='relu'))
# model.add(Dense(8, activation='relu'))
# model.add(Dense(1, activation='sigmoid'))
model = Sequential([
    Input(shape=(8,)),
    Dense(10, activation='relu'),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
], name='MLP_model_Basic')

model.summary()
```

```python
model.compile(
    loss = "binary_crossentropy",
    optimizer = "sgd",
    metrics = ["accuracy"]
)
```

```python
model.fit(
    X,
    Y,
    epochs = 150,
    batch_size = 10,
)
```

```python
loss, accuracy = model.evaluate(X, Y)
print(f"Loss: {loss:.4f}\n\rAccuracy: {accuracy:.4f}")
```

(1) 以 ch5-2-2.py 為基礎，執行訓練測試，結果正確率為 75.78 %
## 模型二 
(2) 以 case(1) 為基礎，增加隱藏層內各層之NN之數目，是否可得較佳之正確率

```python
# model.add(Input(shape=(8,)))
# model.add(Dense(10, activation='relu'))
# model.add(Dense(8, activation='relu'))
# model.add(Dense(1, activation='sigmoid'))
model_1 = Sequential([
    Input(shape=(8,)),
    Dense(15, activation='relu'),
    Dense(10, activation='relu'),
    Dense(1, activation='sigmoid')
], name='MLP_model_1')

model_1.summary()
```

```python
model_1.compile(
    loss = "binary_crossentropy",
    optimizer = "sgd",
    metrics = ["accuracy"]
)
```

```python
model_1.fit(
    X,
    Y,
    epochs = 150,
    batch_size = 10,
)
```

```python
loss, accuracy = model_1.evaluate(X, Y)
print(f"Loss: {loss:.4f}\n\rAccuracy: {accuracy:.4f}")
```

(2) 以 case(1) 為基礎，增加隱藏層內各層之NN之數目， 在隱藏層 15, 10 下，正確率為 73.05% ，顯示結果比原始還差
## 模型三 
(3) 以 case(1) 為基礎，增加隱藏層之層數，是否可得較佳之正確率

```python
model_2 = Sequential([
    Input(shape=(8,)),
    Dense(20, activation='relu'),
    Dense(15, activation='relu'),
    Dense(10, activation='relu'),
    Dense(5, activation='relu'),
    Dense(1, activation='sigmoid')
], name='MLP_model_2')

model_2.summary()
```

```python
model_2.compile(
    loss = "binary_crossentropy",
    optimizer = "sgd",
    metrics = ["accuracy"]
)
```

```python
model_2.fit(
    X,
    Y,
    epochs = 150,
    batch_size = 10,
)
```

```python
loss, accuracy = model_2.evaluate(X, Y)
print(f"Loss: {loss:.4f}\n\rAccuracy: {accuracy:.4f}")
```

(3) 以 case(1) 為基礎，增加隱藏層之層數，正確率為 74.74% 仍沒有比原始好

## 結論
透過本次使用 Keras 建立多層感知器（MLP）進行糖尿病資料集二元分類的實作，我體會到神經網路並非「參數越多、層數越深就一定越準」。在固定 sgd 優化器、batch_size=10 與 epochs=150 的對照實驗中，原始架構（10 -> 8）獲得了 75.78% 的準確率（Loss 為 0.4938）；但當我增加隱藏層神經元數（15 -> 10，共 306 個參數）時，準確率反而降至 73.05%（Loss 0.5217）；將隱藏層加深至四層（20 -> 15 -> 10 -> 5，共 716 個參數）後，準確率也僅為 74.74%（Loss 0.5176），仍未超越基準模型。   

進一步分析原因後，我發現這份資料集僅有 768 筆樣本且未做特徵標準化，各生理指標的數值尺度差異極大，同時，傳統 SGD 優化器收斂慢。當網路盲目加寬或加深時，待優化的權重空間變大，在短短 150 個 Epoch 內反而無法充分收斂至較佳解。這次實驗讓我深刻體會到，實務上做好資料前處理（如標準化）與選對優化器，往往比單純堆疊網路深度與寬度更為關鍵。
