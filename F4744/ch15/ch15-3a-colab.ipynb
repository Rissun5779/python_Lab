```python
from keras_nlp.models import GPT2CausalLMPreprocessor, GPT2CausalLM
import keras
import time

keras.mixed_precision.set_global_policy("mixed_float16")
# GPT-2 的預處理, 為了加速將文字長度從全長 1024 改成 128
preprocessor = GPT2CausalLMPreprocessor.from_preset(
    "gpt2_base_en",
    sequence_length=128,
)
# 載入 GPT-2 模型
gpt2_lm = GPT2CausalLM.from_preset(
    "gpt2_base_en",
    preprocessor=preprocessor
)

# 使用 GPT-2 模型來產生文字
start = time.time()
output = gpt2_lm.generate("My trip to New York was", max_length=200)
print("\nGPT-2 輸出:")
print(output)
end = time.time()
print(f"花費時間: {end - start:.2f}s")
```
